Read From CSV and Write Bronze Table

```python
# Initialization if this would be another notebook
# from bronze_improved import INGESTION_CONFIG

# %sql
# USE CATALOG worskpace;
# USE SCHEMA bronze;
```

```python
for item in INGESTION_CONFIG:
    print(f"Ingesting {item['source']} -> workspace.bronze.{item['table']}")

    df = (
        spark.read
             .option("header", "true")
             .option("inferSchema", "true")
             .csv(item["path"])
    )

    (
        df.write
          .mode("overwrite")
          .format("delta")
          .saveAsTable(f"workspace.bronze.{item['table']}")
    )
```
